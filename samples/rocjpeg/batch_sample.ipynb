#### Python Sample - Batch decode

This sample decodes a batch of multiple JPEG images, passing all the images full path as a list, or loading their data and pass it as a list of memory buffer(s). 

```python
import pyRocJpegDecode.decoder as jdec
import sys
```

Load images

```python
image_paths = ["/opt/rocm/share/rocjpeg/images/mug_400.jpg", "/opt/rocm/share/rocjpeg/images/mug_420.jpg", "/opt/rocm/share/rocjpeg/images/mug_422.jpg"]
batch_size = len(image_paths)
```

Create decoder

```python
device_id = 0
backend = 0

if( jdec.initialize_hip(device_id) == False):
    print(f"Device#: {device_id} not found.")
    sys.exit(1)

decoder = jdec.decoder(device_id, backend)
```

Decode data

```python
data_list = []
for p in image_paths:
    with open(p, 'rb') as in_file:
        data = in_file.read()
        data_list.append(data)
        
image_list = decoder.decode(data_list)
```

Decode files

```python
images = decoder.decode(image_paths)
```
